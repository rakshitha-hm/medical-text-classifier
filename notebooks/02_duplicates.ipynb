```python
import pandas as pd

df = pd.read_csv("../data/mtsamples.csv")
df = df.dropna(subset=["transcription"])
df["medical_specialty"] = df["medical_specialty"].str.strip()
print(len(df))
```

```python
dup_mask = df.duplicated("transcription", keep=False)
print("Rows whose text appears more than once:", dup_mask.sum())

labels_per_text = df[dup_mask].groupby("transcription")["medical_specialty"].nunique()
print(labels_per_text.value_counts())
```

```python
pairs = (df[dup_mask]
         .groupby("transcription")["medical_specialty"]
         .apply(lambda s: " | ".join(sorted(set(s)))))
print(pairs[pairs.str.contains(r"\|")].value_counts().head(15))
```

```python
labels_per_text = df[dup_mask].groupby("transcription")["medical_specialty"].apply(set)
conflicts = labels_per_text[labels_per_text.apply(len) > 1]
print("Notes with conflicting labels:", len(conflicts))
print("...involving Surgery:", conflicts.apply(lambda s: "Surgery" in s).sum())
print("...involving Radiology:", conflicts.apply(lambda s: "Radiology" in s).sum())
```
